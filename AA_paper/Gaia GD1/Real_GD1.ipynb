```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.table import Table

tbl = Table.read("./gd1_tbl_with_memb_prob.mrt", format="ascii.mrt")
print(tbl)
```

```python
tbl
```

```python
dataframe = tbl[["parallax", "phi1", "phi2", "rv", "pm1", "pm2", "memb_prob"]].to_pandas()
dataframe["r"] = (dataframe['parallax'].to_numpy() * u.arcsec).to(u.parsec, equivalencies=u.parallax())
dataframe = dataframe.drop(columns=["parallax"])
cols = ["r", "phi1", "phi2",  "rv", "pm1", "pm2", "memb_prob"]
dataframe = dataframe[cols]
dataframe.head()
```

```python
membership_cut = 0.9
dataframe_to_plot = dataframe[(dataframe["memb_prob"] > membership_cut) & (dataframe["r"] > 0.0 ) & (dataframe["rv"] != 0.0 )]
print("Numnber of stars with membership probability >", membership_cut, ":", len(dataframe_to_plot))
fig = plt.figure(figsize=(20,10))
ax = fig.add_subplot(2,2,1)
scatter = ax.scatter(dataframe_to_plot["phi1"], dataframe_to_plot["phi2"], c=dataframe_to_plot["memb_prob"], cmap="viridis", s=5)
plt.colorbar(scatter, label="Membership Probability")
ax.set_xlabel("$\phi_1$ [deg]")
ax.set_ylabel("$\phi_2$ [deg]")

ax = fig.add_subplot(2,2,2)
ax.hist(dataframe_to_plot["memb_prob"], bins=30)
ax.set_xlabel("Membership Probability")
dataframe_to_plot.head()

ax = fig.add_subplot(2,2,3)
scatter = ax.scatter(dataframe_to_plot["pm1"], dataframe_to_plot["pm2"], c=dataframe_to_plot["memb_prob"], cmap="viridis", s=5)
plt.colorbar(scatter, label="Membership Probability")
ax.set_xlabel("$\mu_{\phi_1}$ [mas/yr]")
ax.set_ylabel("$\mu_{\phi_2}$ [mas/yr]")

ax = fig.add_subplot(2,2,4)
scatter = ax.scatter(dataframe_to_plot["r"], dataframe_to_plot["rv"], c=dataframe_to_plot["memb_prob"], cmap="viridis", s=5)
plt.colorbar(scatter, label="Membership Probability")
ax.set_xlim(-1, 20)
ax.set_xlabel("Distance r [kpc]")
ax.set_ylabel("Radial Velocity rv [km/s]")
```

```python
gd1_member = dataframe_to_plot.drop(columns=["memb_prob"]).to_numpy()

from scipy.stats import gaussian_kde
kde_approx = gaussian_kde(gd1_member.T)
gd1_kde_resampled = kde_approx.resample(size=5000).T
gd1_kde_resampled = gd1_kde_resampled[gd1_kde_resampled[:,0] > 0.0]  # remove negative distances
gd1_kde_resampled = gd1_kde_resampled[:1000]
print("Number of stars after KDE resampling:", len(gd1_kde_resampled))
fig = plt.figure(figsize=(20,5))
ax = fig.add_subplot(1, 3, 1)
ax.scatter(gd1_kde_resampled[:,1], gd1_kde_resampled[:,2], s=5, alpha=0.5)
ax.set_xlabel("$\phi_1$ [deg]")
ax.set_ylabel("$\phi_2$ [deg]")

ax = fig.add_subplot(1, 3, 2)
ax.scatter(gd1_kde_resampled[:,4] * np.cos(gd1_kde_resampled[:,2]), gd1_kde_resampled[:, 5], s=5, alpha=0.5)
ax.set_xlabel("$\mu_{\phi_1}$ [mas/yr]")
ax.set_ylabel("$\mu_{\phi_2}$ [mas/yr]")

ax = fig.add_subplot(1, 3, 3)
ax.scatter(gd1_kde_resampled[:,0], gd1_kde_resampled[:,3], s=5, alpha=0.5)
ax.set_xlabel("Distance r [kpc]")
ax.set_ylabel("Radial Velocity rv [km/s]")
ax.set_xlim(-1, 20)
plt.show()
```
